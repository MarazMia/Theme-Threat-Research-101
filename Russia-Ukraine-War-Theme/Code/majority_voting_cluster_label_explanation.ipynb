```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
import shap
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import sklearn.metrics as metrics
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.preprocessing import normalize
from imblearn.over_sampling import SMOTE
```

```python
plt.rcParams["figure.figsize"] = (8,6)
```

```python
data = pd.read_csv('../Data/Clusters/All_5_Clustering_Algo_Result_10-30-2024.csv')
data.head()
```

```python
print(len(data[data['Majority Voting Label'] == 1]))
print(len(data[data['Majority Voting Label'] == 0]))

print('-'*20)

print(len(data[data['IPQS_Flag'] == 1]))
print(len(data[data['IPQS_Flag'] == 0]))

print('-'*20)

print(len(data[(data['IPQS_Flag'] == 1) & (data['Majority Voting Label'] == 1)]))
print(len(data[(data['IPQS_Flag'] == 0) & (data['Majority Voting Label'] == 0)]))
```

```python
ipqs_ben_index = [i for i in range(len(list(data.IPQS_Flag))) if list(data.IPQS_Flag)[i]==0]
ipqs_mal_index = [i for i in range(len(list(data.IPQS_Flag))) if list(data.IPQS_Flag)[i]==1]

print(ipqs_ben_index)
print(ipqs_mal_index)
print(len(ipqs_ben_index))
print(len(ipqs_mal_index))
```

```python
df = data.drop(columns=['IPQS_Flag'], axis=1)
```

```python
print(len(data))
print(len(df))
```

```python
clf_df = df.select_dtypes(include='number')
clf_df = clf_df[clf_df.columns[:28]]
clf_df['Majority Voting Label'] = df['Majority Voting Label']
clf_df.info()
```

```python
if 'Status Code' in clf_df.columns:
    clf_df = clf_df.drop(columns=['Status Code'], axis=1)
```

```python
X = np.array(normalize(clf_df[clf_df.columns[:-1]]))
Y = np.array(clf_df['Majority Voting Label'])
set(Y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.175, random_state=0, shuffle=True)
```

```python
clf = xgb.XGBClassifier(objective='multi:softprob', max_depth=2, random_state=40, n_estimators=200, learning_rate=0.25, min_child_weight=3, gamma=0.05, colsample_bytree=0.1)
clf.fit(x_train, y_train)
```

```python
predicted = clf.predict(x_test)
print(classification_report(y_test, predicted))
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, predicted)
print("Accuracy:", accuracy)

# Calculate precision
precision = precision_score(y_test, predicted, average='weighted')
print("Precision:", precision)

# Calculate recall
recall = recall_score(y_test, predicted, average='weighted')
print("Recall:", recall)

# Calculate F1 score
f1 = f1_score(y_test, predicted, average='weighted')
print("F1 score:", f1)
```

```python
xgb_cm = metrics.confusion_matrix(y_test, predicted)
unique_classes = ['C_1', 'C_2', 'C_3']
fig, ax = plt.subplots(figsize=(6,4))
sns.heatmap(xgb_cm, annot=True, fmt='d', ax=ax, cmap=plt.cm.Blues,
            cbar=False)
ax.set_xlabel("Pred", fontweight='bold')
ax.set_ylabel("True", fontweight='bold')
ax.set(xticklabels=unique_classes, yticklabels=unique_classes)

# Make tick labels bold
ax.set_xticklabels(unique_classes, fontweight='bold')
ax.set_yticklabels(unique_classes, fontweight='bold')
plt.yticks(rotation=0)
plt.show()
```

```python
explainer = shap.TreeExplainer(clf,X)
shap_values = explainer(X,Y)
shap_values.feature_names = clf_df.columns[:-1]

print(shap_values.shape)
```

```python
import numpy as np
import shap
import matplotlib.pyplot as plt

# Calculate the raw mean (not absolute) SHAP values for each class
mean_shap_values = [
    np.mean(shap_values.values[:, :, class_ind], axis=0) for class_ind in range(shap_values.shape[-1])
]

# Number of top features to display
top_n = 28  # Adjust if you want a different number of top features

# Create a bar plot for each class
fig, axes = plt.subplots(1, 3, figsize=(16, 6), sharey=True)
for class_ind, ax in enumerate(axes):
    # Get the current class mean SHAP values
    class_mean_shap_values = mean_shap_values[class_ind][:top_n]
    class_feature_names = clf_df.columns[:-1][:top_n]  # Adjust the number of features to top_n

    # Sort features based on absolute SHAP values (descending order)
    sorted_indices = np.argsort(np.abs(class_mean_shap_values))[::-1]  # Sort in descending order of absolute value
    sorted_shap_values = class_mean_shap_values[sorted_indices]
    sorted_feature_names = np.array(class_feature_names)[sorted_indices]

    # Define colors: blue for negative, red for positive
    colors = ['#0089FA' if val < 0 else '#FF0051' for val in sorted_shap_values]

    # Plot the sorted SHAP values with the selected colors
    ax.barh(sorted_feature_names, sorted_shap_values, color=colors)

    ax.set_title(f"C_{class_ind+1} - Mean SHAP values")
    ax.set_xlabel("Mean SHAP value")
    ax.invert_yaxis()  # Invert y-axis to show the most important feature at the top

    # Set consistent x-limits for better comparison
    ax.set_xlim([min(sorted_shap_values) * 1.1, max(sorted_shap_values) * 1.1])  # Set x-limits for better scaling

# Add a tight layout and show the plot
plt.tight_layout()
plt.show()
```
